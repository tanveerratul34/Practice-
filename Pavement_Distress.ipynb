```python
# STEP 1: Install ultralytics (needed every new Colab session)
!pip install ultralytics -q

# STEP 2: Mount Google Drive (needed every new Colab session)
from google.colab import drive
drive.mount('/content/drive')

# STEP 3: Load your trained model
from ultralytics import YOLO

model_path = "/content/drive/MyDrive/Colab Notebooks/Pavement Dataset/runs/pavement_distress_detect-2/weights/best.pt"

import os
print("Model exists:", os.path.exists(model_path))

model = YOLO(model_path)

# STEP 4: Run on your test image
# <<< CHANGE THIS to your actual uploaded image path >>>
image_path = "/content/drive/MyDrive/your_folder/your_image.jpg"

result = model.predict(source=image_path, save=True, conf=0.25)

# STEP 5: Print detections
if len(result[0].boxes) == 0:
    print("No distress detected in this image.")
else:
    print(f"Detected {len(result[0].boxes)} object(s):")
    for box in result[0].boxes:
        class_id = int(box.cls)
        class_name = result[0].names[class_id]
        confidence = float(box.conf)
        print(f"  - {class_name} (confidence: {confidence:.2f})")

print(f"\nAnnotated image saved to: {result[0].save_dir}")
```
