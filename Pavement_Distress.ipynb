```python
# ============================================================
# STEP 1: Install and mount
# ============================================================
!pip install ultralytics -q
from google.colab import drive
drive.mount('/content/drive')

from ultralytics import YOLO
import csv

# ============================================================
# STEP 2: Load model
# ============================================================
model_path = "/content/drive/MyDrive/Colab Notebooks/Pavement Dataset/runs/pavement_distress_detect-2/weights/best.pt"
model = YOLO(model_path)

# ============================================================
# STEP 3: Point at the folder containing your 25 chosen images
#   <<< CHANGE THIS to your 25-image folder path in Drive >>>
# ============================================================
image_folder = "/content/drive/MyDrive/Colab Notebooks/Pavement Dataset/PCI manual Cal images"

results = model.predict(source=image_folder, save=True, conf=0.15)

# ============================================================
# STEP 4: Extract box coordinates (in pixels) for each detection
#   and export to a CSV you can open directly in Excel
# ============================================================
output_rows = []

for r in results:
    image_name = r.path.split("/")[-1]
    img_w, img_h = r.orig_shape[1], r.orig_shape[0]  # width, height in pixels

    if len(r.boxes) == 0:
        output_rows.append([image_name, img_w, img_h, "no detections", "", "", "", "", "", ""])
    else:
        for box in r.boxes:
            class_id = int(box.cls)
            class_name = r.names[class_id]
            confidence = float(box.conf)

            # xyxy = [x_min, y_min, x_max, y_max] in pixels
            x_min, y_min, x_max, y_max = box.xyxy[0].tolist()
            box_width_px = x_max - x_min
            box_height_px = y_max - y_min
            box_area_px = box_width_px * box_height_px

            output_rows.append([
                image_name, img_w, img_h, class_name,
                round(confidence, 3),
                round(x_min, 1), round(y_min, 1), round(x_max, 1), round(y_max, 1),
                round(box_width_px, 1), round(box_height_px, 1), round(box_area_px, 1)
            ])

# ============================================================
# STEP 5: Save as CSV - to Drive (with custom name) AND download to PC
# ============================================================
csv_filename = "Automated PCI detection box.csv"

# Save directly to Drive
drive_csv_path = f"/content/drive/MyDrive/Colab Notebooks/Pavement Dataset/{csv_filename}"
with open(drive_csv_path, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow([
        "Image", "Image Width (px)", "Image Height (px)", "Class", "Confidence",
        "x_min", "y_min", "x_max", "y_max",
        "Box Width (px)", "Box Height (px)", "Box Area (px^2)"
    ])
    writer.writerows(output_rows)

print(f"Saved to Drive: {drive_csv_path}")

from google.colab import files
files.download(drive_csv_path)

print(f"\nDownload triggered - check your Downloads folder for '{csv_filename}'")
```

```python
import numpy as np

def get_cdv(tdv, q):
    """Approximates ASTM Fig. X3.27 (Total/Corrected Deduct Value curve)."""
    q = max(1, min(q, 8))
    base = tdv * (1 - 0.05*(q-1))
    cdv = 100 * (1 - np.exp(-base/85))
    return round(min(cdv, 100), 1)

def compute_pci(dv_list):
    dv_list = sorted(dv_list, reverse=True)
    if len([d for d in dv_list if d > 2]) <= 1:
        cdv = sum(dv_list)
        return round(100 - cdv, 1)

    max_cdv = 0
    working = dv_list[:]
    while True:
        hdv = working[0]
        m = min(10, 1 + (9/98)*(100-hdv))
        n_full = int(m)
        frac = m - n_full
        reduced = working[:n_full]
        if frac > 0 and len(working) > n_full:
            reduced.append(working[n_full] * frac)
        tdv = sum(reduced)
        q = len([d for d in working if d > 2])
        cdv = get_cdv(tdv, q)
        max_cdv = max(max_cdv, cdv)
        if q <= 1:
            break
        # actually reduce the smallest qualifying DV to 2.0
        for i in range(len(working)-1, -1, -1):
            if working[i] > 2:
                working[i] = 2.0
                break
    return round(100 - max_cdv, 1)

# ---- Test it against your own manually-verified example ----
# Image 1: alligator cracking DV=75, pothole DV=73 -> you calculated PCI=6 by hand
result = compute_pci([75, 73])
print(result)  # should print 6.0
```

```python
!pip install ultralytics scikit-image -q
from google.colab import drive
drive.mount('/content/drive')

import numpy as np, cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO, SAM
from skimage.filters import sato, apply_hysteresis_threshold
from skimage.morphology import skeletonize

det_path = "/content/drive/MyDrive/Colab Notebooks/Pavement Dataset/runs/pavement_distress_detect-2/weights/best.pt"
img_path = "/content/drive/MyDrive/Colab Notebooks/Pavement Dataset/Test Images/image 1.jpeg"

# ---------------- settings ----------------
CONF_ALLIGATOR = 0.10     # lower, so more of the crack network gets a box
CONF_POTHOLE   = 0.40
MIN_POT_FRAC   = 0.03     # pothole boxes below 3% of image area are dropped
NEST_FRAC      = 0.80     # same-class box mostly inside a larger one is dropped
SHOW_CRACKS    = True     # set False to show only boxes and the pothole mask
HYST_LO, HYST_HI = 85, 97 # crack score percentiles inside the alligator region
MIN_BLOB, MIN_LEN = 80, 40
KEEP = {0: "alligator cracking", 4: "pothole"}

# ---------------- 1. detect ----------------
det = YOLO(det_path)
res = det.predict(img_path, conf=CONF_ALLIGATOR, iou=0.5, classes=list(KEEP), verbose=False)[0]
H, W = res.orig_shape
boxes = res.boxes.xyxy.cpu().numpy()
cls   = res.boxes.cls.cpu().numpy().astype(int)
conf  = res.boxes.conf.cpu().numpy()

ok = np.where(cls == 0, conf >= CONF_ALLIGATOR, conf >= CONF_POTHOLE)
boxes, cls, conf = boxes[ok], cls[ok], conf[ok]
areas = (boxes[:, 2] - boxes[:, 0]) * (boxes[:, 3] - boxes[:, 1])

def inside_frac(a, b):
    ix = max(0, min(a[2], b[2]) - max(a[0], b[0]))
    iy = max(0, min(a[3], b[3]) - max(a[1], b[1]))
    return ix * iy / ((a[2] - a[0]) * (a[3] - a[1]) + 1e-9)

# ---------------- 2. clean boxes (largest first) ----------------
keep = np.zeros(len(boxes), bool)
reason = [""] * len(boxes)
kept_idx = []
for i in np.argsort(-areas):
    if cls[i] == 4 and areas[i] < MIN_POT_FRAC * H * W:
        reason[i] = "too small for a pothole"; continue
    if any(cls[j] == cls[i] and inside_frac(boxes[i], boxes[j]) >= NEST_FRAC for j in kept_idx):
        reason[i] = "nested inside a larger box"; continue
    keep[i] = True; kept_idx.append(i)

print("idx | class | conf | status")
for i in range(len(boxes)):
    print(i, KEEP[cls[i]], f"{conf[i]:.2f}", "kept" if keep[i] else f"dropped ({reason[i]})")

dropped = boxes[~keep]
boxes, cls = boxes[keep], cls[keep]

img = cv2.cvtColor(cv2.imread(img_path), cv2.COLOR_BGR2RGB)

# ---------------- 3. potholes: SAM masks from cleaned boxes ----------------
pot = [i for i, c in enumerate(cls) if c == 4]
pot_mask = np.zeros((H, W), bool)
if pot:
    seg = SAM("sam_b.pt")(img_path, bboxes=boxes[pot].tolist(), verbose=False)[0]
    m = seg.masks.data.cpu().numpy()
    if m.shape[1:] != (H, W):
        m = np.stack([cv2.resize(x, (W, H), interpolation=cv2.INTER_NEAREST) for x in m])
    pot_mask = (m > 0.5).any(axis=0)

# ---------------- 4. alligator: classical crack filter (not a trained model) ----------------
crack_full = np.zeros((H, W), bool)
crack_show = crack_full
if SHOW_CRACKS and (cls == 0).any():
    s = 1024 / max(H, W)
    gray8 = cv2.resize(cv2.cvtColor(img, cv2.COLOR_RGB2GRAY), None, fx=s, fy=s)
    g = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8)).apply(gray8)
    bh = cv2.morphologyEx(g, cv2.MORPH_BLACKHAT,
                          cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (31, 31))).astype(float)
    ridge = sato(g.astype(float) / 255, sigmas=[1.5, 3, 5], black_ridges=True)
    score = 0.5 * ridge / (ridge.max() + 1e-9) + 0.5 * bh / (bh.max() + 1e-9)

    region = np.zeros(g.shape, bool)
    for b, c in zip(boxes, cls):
        if c == 0:
            x1, y1, x2, y2 = (b * s).astype(int)
            region[y1:y2, x1:x2] = True
    avoid = cv2.dilate(cv2.resize(pot_mask.astype(np.uint8), g.shape[::-1],
                                  interpolation=cv2.INTER_NEAREST), np.ones((25, 25), np.uint8)) > 0
    for b in dropped:
        x1, y1, x2, y2 = (b * s).astype(int)
        avoid[y1:y2, x1:x2] = True
    region &= ~avoid

    if region.any():
        lo, hi = np.percentile(score[region], [HYST_LO, HYST_HI])
        crack = apply_hysteresis_threshold(score, lo, hi) & region
        n, lab, st, _ = cv2.connectedComponentsWithStats(crack.astype(np.uint8), 8)
        clean = np.zeros_like(crack)
        for k in range(1, n):
            w, h, a = st[k, cv2.CC_STAT_WIDTH], st[k, cv2.CC_STAT_HEIGHT], st[k, cv2.CC_STAT_AREA]
            if a >= MIN_BLOB and max(w, h) >= MIN_LEN:
                clean |= lab == k
        up = lambda mk: cv2.resize(mk.astype(np.uint8), (W, H), interpolation=cv2.INTER_NEAREST) > 0
        crack_full = up(clean)
        crack_show = cv2.dilate(up(skeletonize(clean)).astype(np.uint8), np.ones((7, 7), np.uint8)) > 0

# ---------------- 5. report ----------------
ft2 = (3.5 / W) * (4.5 / H); sample = 3.5 * 4.5
if pot:
    print(f"pothole mask covers {pot_mask.sum()*ft2/sample*100:.1f}% of the frame")
for b, c in zip(boxes, cls):
    if c == 0:
        print(f"alligator box covers {(b[2]-b[0])*(b[3]-b[1])*ft2/sample*100:.1f}% of the frame")
if SHOW_CRACKS:
    print(f"crack-filter pixels inside alligator boxes: {crack_full.sum()*ft2/sample*100:.2f}% of the frame")

# ---------------- 6. figure ----------------
overlay = img.copy()
overlay[pot_mask] = (0.5 * overlay[pot_mask] + 0.5 * np.array([255, 0, 0])).astype(np.uint8)
overlay[crack_show] = [0, 255, 255]
fig, ax = plt.subplots(1, 2, figsize=(14, 7))
ax[0].imshow(img); ax[0].set_title("Cleaned detections (blue = alligator, red = pothole)")
ax[1].imshow(overlay); ax[1].set_title("Pothole: SAM mask (red) | Alligator: crack filter (cyan)")
for a in ax:
    for b, c in zip(boxes, cls):
        a.add_patch(plt.Rectangle((b[0], b[1]), b[2]-b[0], b[3]-b[1], fill=False,
                                  ec="red" if c == 4 else "dodgerblue", lw=2))
    a.axis("off")
plt.tight_layout(); plt.savefig("final_segmentation_v2.png", dpi=200)
from google.colab import files; files.download("final_segmentation_v2.png")
```
