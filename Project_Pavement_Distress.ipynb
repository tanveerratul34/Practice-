```python
# Setup
!pip install ultralytics -q
from google.colab import drive
drive.mount('/content/drive')

from ultralytics import YOLO
import shutil, os

# Load model
model_path = "/content/drive/MyDrive/Colab Notebooks/Pavement Dataset/runs/pavement_distress_detect-2/weights/best.pt"
model = YOLO(model_path)

# Run detection — saves annotated images locally in Colab first
results = model.predict(
    source="/content/drive/MyDrive/Colab Notebooks/Pavement Dataset/Field Images",
    save=True,
    conf=0.25,
    project="/content/drive/MyDrive/Colab Notebooks/Pavement Dataset/detections",  # saves straight to Drive
    name="field_results"
)

# Print detections
for r in results:
    image_name = r.path.split("/")[-1]
    if len(r.boxes) == 0:
        print(f"\n{image_name}: no distress detected")
    else:
        print(f"\n{image_name}: {len(r.boxes)} detection(s)")
        for box in r.boxes:
            class_id = int(box.cls)
            class_name = r.names[class_id]
            confidence = float(box.conf)
            print(f"  - {class_name} (confidence: {confidence:.2f})")

result_folder = results[0].save_dir
print(f"\nAnnotated images saved to Drive at: {result_folder}")

# Zip the whole results folder and trigger download to PC
zip_path = "/content/field_results"
shutil.make_archive(zip_path, 'zip', result_folder)

from google.colab import files
files.download(f"{zip_path}.zip")

print("\nDownload triggered — check your browser's Downloads folder for field_results.zip")
```
